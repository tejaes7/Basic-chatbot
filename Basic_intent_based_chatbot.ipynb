```python
import json

with open('/content/sample_data/intent.json', 'r') as f:
    intent_data = json.load(f)

print("JSON data loaded successfully:")
print(intent_data)
```

```python
import nltk
import string
from nltk.stem import PorterStemmer


stemmer = PorterStemmer()

def preprocesstext(text):
  words = nltk.word_tokenize(text.lower())
  words = [stemmer.stem(word) for word in words if word not in string.punctuation]
  return " ".join(words)
```

```python
import nltk
nltk.download('punkt_tab', quiet=True)

text=[]
labels = []

for intent in intent_data['intents']:
  for pattern in intent['patterns']:
    text.append(preprocesstext(pattern))

    labels.append(intent['tag'])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidvec = TfidfVectorizer(ngram_range=(1,2))
X = tfidvec.fit_transform(text)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X,labels)
```

```python
import random
def chatbot():
  while True:
    usermsg = input("You:")
    usermsg = preprocesstext(usermsg)
    vec = tfidvec.transform([usermsg])

    intent = model.predict(vec)[0]

    for i in intent_data['intents']:
      if i['tag'] == intent:
        print("Bot:",random.choice(i['responses']))
```

```python
chatbot()
```
